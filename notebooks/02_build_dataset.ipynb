### **Constructor de dataset**
Este notebook une los candidatos extraido con correspondiente etiqueta y genera el dataset final para entrenar el modelo.

#### **Celda 1 – Imports y rutas**

```python
import os, sys
from pathlib import Path
import pandas as pd

sys.path.append(os.path.abspath(".."))
from src.config import DATA_PROCESSED, DATA_LABELS

MASTER = DATA_PROCESSED / "dataset_master.csv"
LABELS = DATA_LABELS / "etiquetas.csv"
OUT    = DATA_PROCESSED / "dataset.csv"

MASTER, LABELS, OUT
```

#### **Celda 2 – Merge**

```python
dm = pd.read_csv(MASTER)
lab = pd.read_csv(LABELS) if LABELS.exists() else pd.DataFrame(columns=["context_hash","label","rationale"])

print("master:", dm.shape, "labels:", lab.shape)

ds = dm.merge(lab[["context_hash","label"]], on="context_hash", how="inner")
ds = ds.dropna(subset=["label"]).reset_index(drop=True)
ds["label"] = ds["label"].astype(int)

print("dataset:", ds.shape)
ds["label"].value_counts()
```

#### **Celda 3 – Guardar**

```python
OUT.parent.mkdir(parents=True, exist_ok=True)
ds.to_csv(OUT, index=False)
print("Guardado:", OUT)
```
